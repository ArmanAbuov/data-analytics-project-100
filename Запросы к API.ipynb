```python
import pandas as pd
import requests
```

```python
response1 = requests.get('https://data-charts-api.hexlet.app/visits?begin=2023-03-01&end=2023-09-01')
if response1.status_code == 200:
#    print(response1.json())
    df1 = pd.DataFrame(response1.json())
    print(df1.head())
    df1.to_csv('test190526.csv')
```
